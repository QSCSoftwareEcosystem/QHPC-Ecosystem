# EQO notebook: H6 QFlow boundary

The H6 workflow is valuable evidence of how an eventual heterogeneous chemistry workflow combines services, but it is **not executable in EQO today**. This notebook is intentionally a review aid, not a run-submission example.

## Evidence-backed H6 cycle 1

H6/STO-3G input → native ExaChem QFlow observe/export → shared amplitude snapshot and three formed-Hamiltonian tasks → scheduler-neutral QIRIS contract replay → native NWQSim plugin once per task → aggregate by task identity → QFlow complete-cycle acceptance → stop before live QFlow amplitude application.

The saved comparison passed a 1e-8 hartree tolerance for all three tasks. It demonstrates the proposed data and ownership boundary, not an admitted ExaChem, QIRIS, or NWQSim service.

## Why there is no run cell

The QFlow task-set export hook and NWQSim plugin are uncommitted development patches; live QIRIS submission, QIR generation, target scheduler admission, amplitude application, and restart equivalence remain unresolved. Compose may show the blueprint, while Runs and the SDK intentionally cannot submit it.

See integrations/qiris-qflow/h6-validation-workflow.md and integrations/qiris-qflow/architecture.md for the fixed fixture and gates.

## FTQC relationship

FTQC is a future branch only. The validated H6 evidence does not emit an OpenQASM artifact. Once a QIRIS-selected backend emits qhpc.quantum-circuit@1, the separately executable FTQC → IQM notebook can become a downstream step: QIRIS-selected kernel → OpenQASM artifact → FTQC import/lowering → FTQC MLIR → admitted target execution.

Until then, do not portray the H6 validation evidence as an FTQC or hardware execution.
