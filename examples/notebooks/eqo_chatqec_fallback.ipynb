# EQO notebook: ChatQEC through EQO

This notebook calls the EQO control plane, never a ChatQEC container, Qdrant instance, model provider, or MCP tool directly. Start EQO Local first with `eqo local up`.

```python
import os

from eqo import EQOClient, render_citations

EQO_ENDPOINT = os.environ.get("EQO_ENDPOINT", "http://127.0.0.1:8081")
eqo = EQOClient.connect(EQO_ENDPOINT)
status = eqo.assistant.status()
status
```

```python
if not status.get("available", False):
    raise RuntimeError("ChatQEC is unavailable. Check `eqo local status` and try again.")

answer = eqo.assistant.ask("What is the surface code?")
print(answer["answer"])
render_citations(answer.get("citations", []))
```

The returned `status` identifies the active EQO-managed ChatQEC mode. The SDK sends no model-provider configuration, corpus location, MCP command, or web-search setting. Citations are rendered as escaped metadata, not executable links.
